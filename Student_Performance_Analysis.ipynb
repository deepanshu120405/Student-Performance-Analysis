```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

plt.style.use('ggplot')
```

```python
df = pd.read_csv('StudentsPerformance.csv')

df.head()
```

```python
df.info()
```

```python
df.describe()
```

```python
df.isnull().sum()
```

```python
df['Average Score'] = (
    df['math score'] +
    df['reading score'] +
    df['writing score']
)/3

df.head()
```

```python
subjects = ['math score',
            'reading score',
            'writing score']

df[subjects].mean().plot(
    kind='bar'
)

plt.title('Average Marks by Subject')
plt.show()
```

```python
plt.figure(figsize=(8,5))

sns.histplot(
    df['Average Score'],
    bins=20,
    kde=True
)

plt.title('Distribution of Average Score')
plt.show()
```

```python
gender_avg = df.groupby(
    'gender'
)['Average Score'].mean()

gender_avg.plot(
    kind='bar'
)

plt.title('Performance by Gender')
plt.show()
```

```python
prep_avg = df.groupby(
'test preparation course'
)['Average Score'].mean()

prep_avg.plot(
kind='bar'
)

plt.title(
'Effect of Test Preparation'
)

plt.show()
```

```python
lunch_avg = df.groupby(
'lunch'
)['Average Score'].mean()

lunch_avg.plot(
kind='bar'
)

plt.title(
'Performance by Lunch Type'
)

plt.show()
```

```python
plt.figure(figsize=(10,5))

sns.barplot(
data=df,
x='parental level of education',
y='Average Score'
)

plt.xticks(rotation=45)

plt.show()
```

```python
corr = df[
[
'math score',
'reading score',
'writing score',
'Average Score'
]
].corr()

plt.figure(figsize=(6,4))

sns.heatmap(
corr,
annot=True,
cmap='coolwarm'
)

plt.show()
```

# INSIGHTS

1. Reading score has highest average marks.
2. Students completing test preparation course perform better.
3. Standard lunch students score higher.
4. Parent education has positive impact on performance.
5. Math, Reading and Writing scores are highly correlated.
